Dictionaries : Dictionaries are unordered collections of items.They store data in key value pairs.Key must be unique and immutable(Ex. Strings , numbers or tuples) , while values can be any type.


```python
# Creating Dictionaries --> there are two ways

# [] -> this is for empty list , () -> this is for empty tuples , {} -> this is for empty sets for dictonary also curly braces , another way is to using the dict class

empty_dist = {}
print(type(empty_dist))

```

```python
# another way is to using the dict class

empty_dist = dict()
empty_dist
```

```python
student = {"name" : "Ram" , "age" : "32" , "grade" : "A+"} # here is a complete dictionary where it has keys name , age , grade,and the values are ram , 32 , A+
print(student)
print(type(student))
```

```python
# what are the most probable mistakes --> make sure that single is always used.

student = {"name" : "Ram" , "age" : "32" , "name" : "A+"} 
print(student)
# here the key is already defined if again the same key comes the value get replaced with the recent value.so make sure that the key is always unique
```

```python
# Accessing Dictionary Elements.
student = {"name" : "Ram" , "age" : "32" , "grade" : "A+"}

# if i want to access the grade 

print(student["grade"])
print(student["age"])   # this is one way

# Another way is default get() method in the dictionary
print(student.get('grade'))
print(student.get("last_name")) # what if we want to access the key which is not in the list

# if we want to give some default value for the key which is not available in the dictionaary

print(student.get("last_name" , "Not Avaialble"))
```

```python
# Modifying Dictionary Elements --> dictionaries are mutable so you can update or delete elements. but the key needs to be unique at all the time
# 1. updating the age value 2.adding one new key 3. Deleting any key

print(student) 
student["age"] = 33
print(student)

# 2. adding the new key value pair

student["address"] = "India"
print(student)

# 3. Deleting the key grade --> we use `del` key that is available in python

del student['grade']
print(student)




```

```python
## Common dictionary methods

# 1. if you want to access the all keys and values

keys = student.keys()   # this .keys() method get all the keys
print(keys)
values = student.values() # this .values() method get all the values
print(values) 

#2. Key value pairs in the form of dictionary items

items = student.items() # this .items() get all the key value pairs as the dictionary items ,
print(items) #  basicaly you get in the form of list of tuples
```

```python
# Shallow copy

# lets create a variable called student_copy and assign student dictionry to it

student_copy = student
print(student)
print(student_copy)
  # in the both print statements we get the same.

```

```python
# now lets take name value and assign to some thing else
student["name"] = "Mohan"

print(student)
print(student_copy) # to overcome this
```

```python
student_copy_1 = student.copy()  # shallow copy --> by writing the .copy() we r another memory location
print(student_copy_1)
print(student)
```

```python
student["name"] = "Shiva"
print(student)
print(student_copy_1)
```

```python
# iterating over dictionaries --> we can use loops to iterate 

# 1. iterating over keys

for keys in student.keys():
    print(keys)

# similarly iterate over values    

for values in student.values():
    print(values)

# Iterate over kev value pairs which is items(nothing but list of tuples)

for items in student.items():
    print(items)

for key,value in student.items():
    print(f"{key}:{value}")



```

```python
# Nested Dictionaries --> dictionary inside another dictionary

students = {
    "student_1" : {"name" : "Alex" , "age" : 25},
    "student_2" : {"name" : "Bob" , "age" : 28},
    "student_3" : {"name" : "Charle" , "age" : 19}
}

print(students)


```

```python
# if you want to access the 2nd student age

print(students["student_2"]["age"])
```

```python
print(students.items())
```

```python
# if i want to iterate over the nested dictionaries

for student_no , student_info in students.items():
    print(f"{student_no}:{student_info}")
    for key,value in student_info.items():    # u can iterate over the inside also
        print(f"{key}:{value}")

```

```python
# Dictionary Comprehension
# 1. find out th square of numbers
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
#2. find out the square of the only even numbers --> conditional dictionary comprehension
even_squares = {x:x**2 for x in range(10) if x%2==0}
print(even_squares)
```

```python
# Practical exmaples

# use a dictionary to count frequency of the elements in the list
numbers = [1,2,2,3,3,3,4,4,4,4]  # so here output should be 1 is present in 1 time , 2 is present in 2times like that

count_frequency = {}

for num in numbers:
    if num in count_frequency:
        count_frequency[num]+=1
    else:
        count_frequency[num] = 1
print(count_frequency)

 
```

```python
# Merge two dictionaries in to one  -->

dict_1 = {"a" : 1 , "b" : 2}
dict_2 = {"b" : 3 , "c" : 4}

merged_dict = {**dict_1 , **dict_2}  # this one is more about when we discuss about functions
print(merged_dict)

```
