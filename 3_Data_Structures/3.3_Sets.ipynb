Sets

Sets are built in datatype in python used to store collections of unique items.they are unordered which means they donot follow the specific order and they donot allow duplicate elements.Sets are useful for membership tests, eliminating duplicate entries,and perfroming mathematical operations like Union, Intersection , Difference and symetric difference 

```python
# How to create Sets ? --> by using the curly braces {}
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
# i can create a empty set as well
my_empty_set = set() # i can create a function like this
print(type(my_empty_set))
```

```python
# inside this empty set i can give tuples , lists
my_set = set([1,2,3,45]) 
print(my_set)
```

```python
# Now write some duplicate elements and see
my_set = set([1,2,3,45,1,5,2]) 
print(my_set)  # here it will automatically remove the duplicates
```

```python
# Basic Sets operations

#1. Adding and Removing the elements
my_set

# now i want to add some elements --> to do that use .add()
my_set.add(7)
print(my_set)

# similarly if i want to add the duplicate element    --> obviously it will ignore the element
my_set.add(7)
print(my_set)  

# 2. Removing the elements --> .remove()

my_set.remove(45)
print(my_set) 

# what if i want to remove the element which is not present at all in the set

my_set.remove(98)
print(my_set)



```

```python
# so if you want like if the element is there then remove or else just let it be with out error in this case use .discard()

my_set.discard(45)
print(my_set)
```

```python
# removing the random element using the  pop()
removed_ele = my_set.pop()
print(removed_ele)  # after executing this statement for 5 times you will be left with no element. so if you try to  run this statement again you get the error.
# print(my_set) 
```

```python
# Set membership test. --> wether that particular ele is memeber of this test or not

my_set = {1,2,3,4,5,6,7}
# how to check ?? using "in".

print(3 in my_set)
print(10 in my_set)


```

```python
# Mathematical Operations --> 
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

# 1. Union --> it will combine all the sets

union_set = set1.union(set2)
print(union_set)

# 2. Intersection --> it will give the common elements that are present in the both sets

intersection_set = set1.intersection(set2) 
print(intersection_set)

# intesection_update() --> 

set1.intersection_update(set2)
print(set1)   ## what ever we did the intersection those elements and will be update in the set1  
print(set2)

```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

# Difference --> 

print(set1.difference(set2))  # this means that we remove the common elements between the two in the set1 and print the remain elements.
```

```python
# Symmetric difference  --> it will remove the common elements between the sets and combine the unique elements between them

print(set1.symmetric_difference(set2))

```

```python
# Sets Methods

set1={1,2,3}
set2={3,4,5}

# what is subset 

print(set1.issubset(set2))

# what is supersubset

print(set1.issuperset(set2))  # It should be like one set should contain all the elements  set1={1,2,3,4,5} set2={3,4,5} now here you will get the true here.
```

```python
# first lets create list 
lst = [1,2,2,3,4,5,5,6,6,7]
# now if i want to remove the duplicate in the list --> we can use the set

set(lst) 
```

```python
# counting unique words in text

text = "In this tutorial we are discussing about the sets in python  the  python"

# to do that 1st we need to split and then use the set to remove duplicate or get the unique and then count it up using len fuction

split_words = text.split()

unique_words = set(split_words)
print(unique_words)

#len(text)

# len(split_words)

print(len(split_words))
print(len(unique_words)) 
```
