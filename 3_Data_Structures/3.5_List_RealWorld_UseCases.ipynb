Ex.1 : Manage a todo list

.Create a todo list to keep track of tasks


```python
to_do_list = ["Clean the house" , "Buy groceries" , "Pay Bills"]

# 1. after some time i want to add some other task to todo list

to_do_list.append("Shedule a meeting")
to_do_list.append("Go to the zym")

# 2. now i completed one task after completing it i just want to remove it
to_do_list.remove("Clean the house")

#3. Now checking if the particular task is in the list or not

if "Pay Bills" in to_do_list:
    print("Don't forget to pay the bills")

print("to do list remaining")
for task in to_do_list:
    print(f"-{task}")
```

Example:2 --> Organising student marks
create a list to store and calculate average marks for students


```python
# Organising student grades
marks =[85,92,78,90,88]

# Adding a new marks
marks.append(95)

print(f"After adding one more : {marks}")

# Calculating the average marks
average_marks = sum(marks)/len(marks)
print(f"Average Marks : {average_marks:.2f}") 

#finding the highest and lowest marks
max_marks = max(marks)
min_marks = min(marks)

print(f"Maximum marks : {max_marks}")
print(f"Minimum marks : {min_marks}")
```

Ex:3 --> Managing the inventory
Use a list to manage inventory items in a store



```python
# Managing an inventory
inventory = ["Apple","Banana","Citrus","Oranges"]

# Adding a new item
inventory.append("Grapes")

# Remove an item that is out of stock

inventory.remove("Banana")

# Checking if item is present or not

item=input("Enter a fruit name to check")
if item in inventory:
    print(f"{item} are in stock")
else:
    print(f"{item} is out of stock")

# Printing the inventory
print("Printing the inventory")
for item in inventory:
    print(f"- {item}")    

```

Ex:4 --> Collecting the user feedback 
    
    .Use a list to collect and analyze user feedback.


```python
# Collecting the user feedback
feedback=["Great service!" , "Very Satisfied" , "Could be better" , "Excellent Experience"]

# Adding a new feedback
feedback.append("Not happy with the service")

#Counting the specific feedback
positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())
print(f"Positive feedback count is : {positive_feedback_count}")

#printing all feedback
print("User Feedback")
for comment in feedback:
    print(f"- {comment}")
```
