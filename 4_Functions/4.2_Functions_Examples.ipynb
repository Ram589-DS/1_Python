```python
# 1. Temperature Convesion

def temp_conversion(temp,unit):
    """This function converts Celsius to Fahrenheit"""
    if unit == 'C':
        return temp * 9/5 + 32   # this is formula for converting temp celsius to fahrenheit
    elif unit == 'F':
        return (temp-32)*5/9    # Fahrenheit to Celsius
    else:
        return None             # if i don't give any parameters

print(temp_conversion(25 , 'C'))
print(temp_conversion(84,'F'))
print(temp_conversion(0,'F')) 
```

```python
# 2. Password Strength checker --> it should contain min of 6 character including one special char and Capital letter and one numeric 

def is_strong_password(password):
    """This function is to check if password is strong or not"""
    if len(password)<7:
        return print("it is weak passowrd")
    if not any(char.isdigit() for char in password):
        return False
    if not any( char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any( char in '!@#$%^&*()_+=|}{}[];:<>?"' for char in password):
        return False
    return True 

print(is_strong_password("Pinninlk"))
print(is_strong_password("StrongPassword!1"))
```

```python
# 3. Calculate total cost of items in a shoping cart

def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost+=item['price']*item['quantity']
    return total_cost




# Example of cart data ---> its a dictionary which is key values pairs 

cart=[
    {'name' : 'Apple' , 'price': 2 , 'quantity': 4},
    {'name' : 'Banana' , 'price': 1 , 'quantity': 5},
    {'name' : 'Citrus' , 'price' : 1 , 'quantity' : 6}  
]

# call the function
print(calculate_total_cost(cart))
```
