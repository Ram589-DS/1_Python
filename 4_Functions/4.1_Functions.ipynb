Introductions to Functions

Function is a block of code that performs a specific task.Functions help in organizing code,reusing the code and improoving readability.




```python
# Syntax --> function starts with `def` keyword and function_name() parameters inside if any
def function_name(parameters):
    """Docstring"""   # it is to use describe what this function does
    # function body
    return expression # it is not necessary some functions may not have any expression returned.but some have.
```

```python
# why do we require functions ---> let's create a num and check whether it is even or odd

num = 24
if(num%2 == 0):
    print("Yes! it is a even num")
else:
    print("No! it is not a even num")

# now if i want to reuse this peice of code in hundered places where i need to check whether the num is even or not.so copy this code and use that in 100 places.it will be bit tedious.so the better way of reusing the entire code with the help of functions.so the entire write inside the function and where ever you need you just call it.

```

```python
def even_or_odd(num):
    """This function finds whether the num is even or odd"""
    if num%2==0:
        print("Yes! Num is even num")
    else:
        print("No! its not a even num")

# Now call the function

even_or_odd(25)  
even_or_odd(12)      
 
```

```python
# function with multiple parameters

def add(a,b):
    """Adding two numbers"""
    c=a+b
    return c; # instead of this you can directly write return a+b;

result = add(2,5)
print(f"sum of two num is : {result}")
```

```python
# Default Parameter

def greet(name):
    print(f"Hello {name} Welcom to the world")

greet("Ram")
```

```python
# if you execute the above function with out parameter you get the error
def greet(name):
    print(f"Hello {name} Welcom to the world" )
    
greet()
```

```python
# here the default parameter take it as guest.
def greet(name = "Guest"):
    print(f"Hello {name} Welcom to the world" )

greet()
# if you give the name
greet("Ram")
```

```python
# Variable Length Arguments --> there are two types
# 1. Postional arguments 2. Key word arguments

# lets create function with many parameters like more and more def print_num(n1,n2,n3,n4,n5....) , instead of all write each and every parameter we can write (*args)
   
def print_num(*args):     # as we know this parameter has a lot of numbers so we need to use the for in loop
    for number in args:      
        print(number)
```

```python
print_num(1,2,3,4,5,"Ram",2.5,True)  # so these parameters are called positonal arguments
```

```python
# Key words arguments --> how it is different from postional arguments it will be denoted by double star ** and one more imp is all the arguments present is in the form of `key value pairs`. so if it is key value pair we can access how we access the Dictionaries.

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")

# now how we do provide the key and value
print_details(name="Ram" , age=32 , profession="Analyst" , country="India")

```

```python
# we can combine both postional and key arguments

def print_details(*args , **kwargs):
    for val in args:
        print(f"Postinal arguments : {val}")
    for key,value in kwargs.items():
        print(f"Key word arguments :{key} : {value}")     

print_details(1,2,3,4,5,"Ram",2.5,True ,name="Ram" , age=32 , profession="Analyst" , country="India")


```

```python
# return statemnt ---> here our function returns something,after the entire statment is executed.

def multiply(a,b):
    return a*b

result = multiply(5,2)
print(f"Multiplication of two num is : {result}")

# instead of writing two lines you can directly call the function

multiply(5,2)  # both give the same result.
```
